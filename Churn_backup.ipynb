```python
# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Columns to analyze
columns = [
    "Age",
    "Tenure",
    "Support Calls",
    "Payment Delay",
    "Total Spend"
]

# Calculate statistics
print("CUSTOMER CHURN DATASET: SUMMARY STATISTICS")
print("=" * 50)

for col in columns:
    print(f"\n{col}")
    print("-" * 30)
    print("Average:", round(df[col].mean(), 2))
    print("Median:", round(df[col].median(), 2))
    print("Minimum:", df[col].min())
    print("Maximum:", df[col].max())
    print("Standard Deviation:", round(df[col].std(), 2))

# Overall summary table
summary = df[columns].agg(
    ["count", "mean", "median", "std", "min", "max"]
).T

summary.columns = [
    "Count", "Average", "Median",
    "Standard Deviation", "Minimum", "Maximum"
]

print("\n\nSUMMARY TABLE")
print(summary.round(2))
```

```python
# 0 = retained, 1 = churned
comparison = df.groupby("Churn")[
    ["Age", "Tenure", "Support Calls", "Payment Delay", "Total Spend"]
].agg(["mean", "median", "std"])

print(comparison.round(2))
```
